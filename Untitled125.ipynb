```python
import os
import json
import pickle
from datetime import datetime, timezone

import h5py
import yaml
import numpy as np
import pandas as pd

from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# ============================================================
# 1. PROJECT PATHS
# ============================================================

PROJECT_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Lung Cancer Incidence Trend Analysis"
)

CSV_PATH = os.path.join(
    PROJECT_DIR,
    "RS_Session_258_AU_1555_C_to_D.csv"
)

H5_PATH = os.path.join(
    PROJECT_DIR, "lung_cancer_model.h5"
)

PKL_PATH = os.path.join(
    PROJECT_DIR, "lung_cancer_model.pkl"
)

YAML_PATH = os.path.join(
    PROJECT_DIR, "config.yaml"
)

JSON_PATH = os.path.join(
    PROJECT_DIR, "model_metadata.json"
)

os.makedirs(PROJECT_DIR, exist_ok=True)

# ============================================================
# 2. LOAD DATASET
# ============================================================

if not os.path.exists(CSV_PATH):
    raise FileNotFoundError(
        f"Dataset not found: {CSV_PATH}"
    )

df = pd.read_csv(CSV_PATH, encoding="utf-8-sig")

print("\nOriginal Dataset:")
print(df.to_string(index=False))

print("\nDataset Shape:", df.shape)
print("\nColumns:", df.columns.tolist())

# ============================================================
# 3. EXTRACT YEAR-WISE LUNG CANCER CASES
# ============================================================

# Expected source structure:
# Year,2020,2021,2022,Increase % (2020-2022)
# Estimated Incidence of Lung cancer Cases in India,
# 98278,100792,103371,5.2

year_columns = [
    col for col in df.columns
    if str(col).strip().isdigit()
    and len(str(col).strip()) == 4
]

if not year_columns:
    raise ValueError(
        "No four-digit year columns found in the CSV."
    )

if len(df) != 1:
    raise ValueError(
        "Expected one incidence row. "
        "Please select the correct row before training."
    )

records = []

for col in year_columns:
    year = int(str(col).strip())
    value = pd.to_numeric(df[col].iloc[0], errors="coerce")

    if pd.notna(value):
        records.append({
            "Year": year,
            "Cases": float(value)
        })

data = pd.DataFrame(records)
data = data.sort_values("Year").reset_index(drop=True)

if len(data) < 2:
    raise ValueError(
        "At least two historical years are required."
    )

if data["Year"].duplicated().any():
    raise ValueError("Duplicate years detected.")

print("\nProcessed Dataset:")
print(data.to_string(index=False))

# ============================================================
# 4. FEATURE PREPARATION
# ============================================================

# Center years to improve numerical stability.

BASE_YEAR = int(data["Year"].min())

X = (
    data["Year"].to_numpy(dtype=float) - BASE_YEAR
).reshape(-1, 1)

y = data["Cases"].to_numpy(dtype=float)

# ============================================================
# 5. TRAIN LINEAR REGRESSION MODEL
# ============================================================

model = LinearRegression()
model.fit(X, y)

train_predictions = model.predict(X)

print("\nModel Training Completed!")

print("Coefficient:", model.coef_[0])
print("Intercept:", model.intercept_)

# ============================================================
# 6. TRAINING FIT METRICS
# ============================================================

mae = mean_absolute_error(y, train_predictions)
mse = mean_squared_error(y, train_predictions)
rmse = np.sqrt(mse)
r2 = r2_score(y, train_predictions)

mape = float(
    np.mean(np.abs((y - train_predictions) / y))
    * 100
) if np.all(y != 0) else None

metrics = {
    "MAE": float(mae),
    "MSE": float(mse),
    "RMSE": float(rmse),
    "R2_Score": float(r2),
    "MAPE_Percent": mape
}

print("\nTraining Fit Metrics:")
for key, value in metrics.items():
    print(f"{key}: {value}")

print(
    "\nWARNING: These are in-sample training metrics, "
    "not independent test accuracy."
)

# ============================================================
# 7. FUTURE INCIDENCE PROJECTIONS
# ============================================================

last_year = int(data["Year"].max())

future_years = np.arange(
    last_year + 1,
    2031
)

if len(future_years) > 0:
    future_X = (
        future_years.astype(float) - BASE_YEAR
    ).reshape(-1, 1)

    future_predictions = model.predict(future_X)

    # Case counts cannot be negative.
    future_predictions = np.maximum(
        future_predictions, 0
    )

    future_predictions = np.rint(
        future_predictions
    ).astype(int)
else:
    future_predictions = np.array([], dtype=int)

future_df = pd.DataFrame({
    "Year": future_years,
    "Predicted_Cases": future_predictions
})

print("\nExperimental Future Projections:")
print(future_df.to_string(index=False))

# ============================================================
# 8. SAVE TRAINED MODEL AS PKL
# ============================================================

# Save the model together with its year transformation.

model_package = {
    "model": model,
    "base_year": BASE_YEAR,
    "feature_name": "Year",
    "target_name": "Cases",
    "model_type": "LinearRegression",
    "training_years": data["Year"].tolist()
}

with open(PKL_PATH, "wb") as file:
    pickle.dump(model_package, file)

print("\nPKL Model Saved:", PKL_PATH)

# ============================================================
# 9. SAVE MODEL PARAMETERS AS H5
# ============================================================

# H5 stores numeric parameters, not a Keras model.

with h5py.File(H5_PATH, "w") as h5file:

    model_group = h5file.create_group("model")

    model_group.create_dataset(
        "coefficients",
        data=np.asarray(model.coef_, dtype=float)
    )

    model_group.create_dataset(
        "intercept",
        data=float(model.intercept_)
    )

    model_group.create_dataset(
        "base_year",
        data=BASE_YEAR
    )

    data_group = h5file.create_group("training_data")

    data_group.create_dataset(
        "years",
        data=data["Year"].to_numpy(dtype=int)
    )

    data_group.create_dataset(
        "cases",
        data=y
    )

    data_group.create_dataset(
        "fitted_cases",
        data=train_predictions
    )

    metrics_group = h5file.create_group("metrics")

    for key, value in metrics.items():
        if value is not None:
            metrics_group.create_dataset(
                key,
                data=value
            )

    h5file.attrs["project"] = (
        "AI-Based Lung Cancer Incidence Trend Analysis"
    )

    h5file.attrs["model_type"] = "LinearRegression"

    h5file.attrs["created_at"] = (
        datetime.now(timezone.utc).isoformat()
    )

print("H5 File Saved:", H5_PATH)

# ============================================================
# 10. SAVE YAML CONFIGURATION
# ============================================================

config = {
    "project": {
        "name": "AI-Based Lung Cancer Incidence Trend Analysis",
        "description": (
            "Historical incidence analysis and "
            "exploratory linear trend projections"
        ),
        "country": "India",
        "disease": "Lung Cancer"
    },

    "dataset": {
        "file_name": os.path.basename(CSV_PATH),
        "total_records": int(len(data)),
        "start_year": int(data["Year"].min()),
        "end_year": int(data["Year"].max()),
        "features": ["Year"],
        "target": "Cases"
    },

    "model": {
        "algorithm": "LinearRegression",
        "base_year": BASE_YEAR,
        "training_type": "Full Historical Dataset",
        "independent_test_set": False,
        "random_state": None
    },

    "forecast": {
        "start_year": int(last_year + 1),
        "end_year": 2030,
        "type": "Exploratory Extrapolation"
    },

    "output_files": {
        "h5": os.path.basename(H5_PATH),
        "pkl": os.path.basename(PKL_PATH),
        "yaml": os.path.basename(YAML_PATH),
        "json": os.path.basename(JSON_PATH)
    }
}

with open(YAML_PATH, "w", encoding="utf-8") as file:
    yaml.safe_dump(
        config,
        file,
        sort_keys=False,
        allow_unicode=True
    )

print("YAML File Saved:", YAML_PATH)

# ============================================================
# 11. SAVE JSON METADATA
# ============================================================

first_cases = float(data["Cases"].iloc[0])
last_cases = float(data["Cases"].iloc[-1])

growth_percent = (
    (last_cases - first_cases) / first_cases * 100
    if first_cases != 0 else None
)

metadata = {
    "project_name": (
        "AI-Based Lung Cancer Incidence Trend Analysis"
    ),

    "created_at": datetime.now(
        timezone.utc
    ).isoformat(),

    "dataset_information": {
        "file_name": os.path.basename(CSV_PATH),
        "total_years": int(len(data)),
        "years": data["Year"].astype(int).tolist(),
        "cases": data["Cases"].astype(int).tolist()
    },

    "statistics": {
        "minimum_cases": float(data["Cases"].min()),
        "maximum_cases": float(data["Cases"].max()),
        "average_cases": float(data["Cases"].mean()),
        "standard_deviation": float(
            data["Cases"].std(ddof=0)
        ),
        "overall_growth_percent": growth_percent
    },

    "model_information": {
        "algorithm": "LinearRegression",
        "coefficient": float(model.coef_[0]),
        "intercept": float(model.intercept_),
        "base_year": BASE_YEAR
    },

    "training_metrics": metrics,

    "evaluation_note": (
        "Metrics are calculated on training observations. "
        "They do not measure out-of-sample performance."
    ),

    "future_predictions": [
        {
            "year": int(year),
            "predicted_cases": int(cases)
        }
        for year, cases in zip(
            future_years, future_predictions
        )
    ],

    "limitations": [
        "Only three historical observations are available.",
        "No independent validation dataset is available.",
        "Future projections assume a continuing linear trend.",
        "Results should not be used for clinical or "
        "public health decisions."
    ]
}

with open(JSON_PATH, "w", encoding="utf-8") as file:
    json.dump(
        metadata,
        file,
        indent=4,
        ensure_ascii=False
    )

print("JSON File Saved:", JSON_PATH)

# ============================================================
# 12. VERIFY GENERATED FILES
# ============================================================

print("\n" + "=" * 60)
print("GENERATED FILES")
print("=" * 60)

output_files = [
    H5_PATH,
    PKL_PATH,
    YAML_PATH,
    JSON_PATH
]

for path in output_files:
    if os.path.exists(path):
        size_kb = os.path.getsize(path) / 1024
        print(
            f"SUCCESS: {os.path.basename(path)} "
            f"({size_kb:.2f} KB)"
        )
    else:
        print(f"FAILED: {path}")

# ============================================================
# 13. TEST SAVED PKL MODEL
# ============================================================

with open(PKL_PATH, "rb") as file:
    loaded_package = pickle.load(file)

loaded_model = loaded_package["model"]
loaded_base_year = loaded_package["base_year"]

test_year = 2028

test_feature = np.array([
    [test_year - loaded_base_year]
], dtype=float)

test_prediction = loaded_model.predict(test_feature)[0]

print("\nSaved Model Test:")
print("Year:", test_year)
print(
    "Estimated Cases:",
    max(0, round(float(test_prediction)))
)

# ============================================================
# 14. TEST SAVED H5 PARAMETERS
# ============================================================

with h5py.File(H5_PATH, "r") as file:

    coefficient = float(
        file["model/coefficients"][0]
    )

    intercept = float(
        file["model/intercept"][()]
    )

    base_year = int(
        file["model/base_year"][()]
    )

    h5_prediction = (
        coefficient * (test_year - base_year)
        + intercept
    )

print("\nH5 Parameter Test:")
print(
    "Estimated Cases:",
    max(0, round(h5_prediction))
)

assert np.isclose(
    test_prediction,
    h5_prediction
), "H5 and PKL predictions do not match!"

print("\n" + "=" * 60)
print("ALL FOUR FILES GENERATED SUCCESSFULLY!")
print("=" * 60)
```
